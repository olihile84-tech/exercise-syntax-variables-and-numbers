```python
import pandas as pd
import numpy as np

data = {
    'country': ['Italy', 'France', 'Spain', 'Australia', 'US', 'Italy', 'France', 'Spain'],
    'variety': ['Merlot', 'Chardonnay', 'Tempranillo', 'Shiraz', 'Pinot Noir', 'Pinot Grigio', 'Merlot', 'Garnacha'],
    'points': [90, 95, 88, 96, 92, 91, 93, 89],
    'price': [25.0, 40.0, None, 50.0, 35.0, None, 45.0, 20.0],
    'region_1': ['Tuscany', None, 'Rioja', 'Victoria', 'Oregon', None, 'Burgundy', 'Aragon'],
    'taster_name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Charlie', 'Alice', 'Bob']
}

reviews = pd.DataFrame(data)
reviews
```

```python
varieties = reviews.variety
print(varieties)
```

```python
first_five = reviews.iloc[0:5]
print(first_five)
```

```python
df = reviews.loc[[0,2,4], ['country','points']]
print(df)
```

```python
mean_points = reviews.points.mean()
median_points = reviews.points.median()
print(mean_points)
print(median_points)
```

```python
unique_countries = reviews.country.nunique()
print(unique_countries)
```

```python
reviews_per_taster = reviews.groupby('taster_name').taster_name.count()
print(reviews_per_taster)
```

```python
price_extremes = reviews.groupby('variety')['price'].agg(['min','max'])
print(price_extremes)
```

```python
n_missing_prices = reviews.price.isnull().sum()
print(n_missing_prices)
```

```python
filled_region = reviews.region_1.fillna('Unknown')
print(filled_region)
```

```python
renamed = reviews.rename(columns={'region_1':'region'})
print(renamed)
```
