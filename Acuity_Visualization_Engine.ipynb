```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
trades = np.random.normal(loc=0.2,scale=1.5, size=100)

initial_balance = 10000
risk_per_trade = 100

win_rate = np.sum(trades>0)/ len(trades)* 100
expectancy = np.mean(trades)
cumlative_equity_curve = initial_balance + np.cumsum(trades * risk_per_trade)

print(f"Win Rate: {win_rate:.2f}%")
print(f"Expectancy: {expectancy:.2f}")
print(f"Balance: ${cumlative_equity_curve[-1]:,.2f}")

plt.plot(cumlative_equity_curve)
plt.title("Equity Curve - 100 Trades")
plt.xlabel("Trade Number")
plt.ylabel("Account Balance in USD")
plt.show()
```

```python
import matplotlib.pyplot as plt

np.random.seed(100)

strategy_a = np.cumsum(np.random.normal(loc=0, scale=1, size=100))
strategy_b = np.cumsum(np.random.normal(loc=0, scale=1, size=100))

plt.plot( strategy_a, color='blue', linestyle='--', label="Strategy A")
plt.plot(strategy_b, color='green', linestyle='--', label="Strategy B")
plt.title("Strategy A vs StrategyB: Equity Curve Comparison")
plt.xlabel("Trade Number")
plt.ylabel("Account Balance")
plt.grid(True)
plt.legend()
plt.show()
```
