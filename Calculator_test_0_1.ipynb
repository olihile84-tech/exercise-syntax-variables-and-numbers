```python
def calculate_rr(entry, stop_loss, take_profit):
  risk = abs(entry-stop_loss)
  reward = abs (take_profit-entry)
  rr= reward/risk
  print ("Risk:", risk)
  print ("Reward:", reward)
  print ("RR:", round(rr, 3))
```

```python
calculate_rr(1.2000,1.1950,1.2150)
```

```python
def calculate_pip(entry, stop_loss, take_profit):
  pip_risk = abs(entry-stop_loss)*10000
  pip_reward = abs (take_profit-entry)*10000
  rr= pip_reward/pip_risk
  print ("Pip Risk:", pip_risk)
  print ("Pip Reward:", pip_reward)
  print ("RR:", round(rr, 3))
```

```python
calculate_pip(1.2000,1.1950,1.2150)
```
