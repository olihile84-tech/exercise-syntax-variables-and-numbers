```python
import pandas as pd
import numpy as np

data = {
    'country': ['Italy', 'France', 'Spain', 'Australia', 'US', 'Italy', 'France'],
    'variety': ['Merlot', 'Chardonnay', 'Tempranillo', 'Shiraz', 'Pinot Noir', 'Pinot Grigio', 'Merlot'],
    'points': [90, 95, 88, 96, 92, 91, 93],
    'price': [25.0, 40.0, None, 50.0, 35.0, None, 45.0],
    'region_1': ['Tuscany', None, 'Rioja', 'Victoria', 'Oregon', None, 'Burgundy']
}

reviews = pd.DataFrame(data)
reviews
```

```python
dtype = reviews.points.dtype

print(dtype)
```

```python
points_strings = reviews.points.astype(str)
print(points_strings)
```

```python
n_missings_price = reviews.price.isnull().sum()
print(n_missings_price)
```

```python
reviews_per_region = reviews.region_1.fillna('Unknown').value_counts()
print(reviews_per_region)
```
