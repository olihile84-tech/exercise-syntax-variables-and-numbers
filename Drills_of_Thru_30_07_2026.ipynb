```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
returns_a = np.random.choice([-1,1], size=100, p=[0.4, 0.6])
equity_a = np.cumsum(returns_a)

np.random.seed(2)
returns_b = np.random.choice([-1, 1], size=100, p=[0.4, 0.6])
equity_b = np.cumsum(returns_b)

plt.figure(figsize=(10, 6))
plt.plot(equity_a, 'b-', label='Strategy A', linewidth=2)
plt.plot(equity_b, 'g--', label='Strategy B', linewidth=2)
plt.title('Cumulative Equity Curves: Strategy A vs Strategy B')
plt.xlabel('Trade Number')
plt.ylabel('Cumulative Equity')
plt.grid(True)
plt.legend()
plt.show()
```
