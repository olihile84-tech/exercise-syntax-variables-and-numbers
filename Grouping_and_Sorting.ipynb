```python
import pandas as pd

data = {
    'country': ['Italy', 'France', 'Spain', 'Australia', 'US', 'Italy', 'France'],
        'variety': ['Merlot', 'Chardonnay', 'Tempranillo', 'Shiraz', 'Pinot Noir', 'Pinot Grigio', 'Merlot'],
            'points': [90, 95, 88, 96, 92, 91, 93],
                'price': [25.0, 40.0, 15.0, 50.0, 35.0, 20.0, 45.0],
                    'taster_name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Charlie', 'Alice']
                    }

reviews = pd.DataFrame(data)
reviews
```

```python
reviews_written = reviews.groupby('taster_name').taster_name.count()
print(reviews_written)
```

```python
best_rating_per_price = reviews.groupby('price')['points'].max().sort_index()
print(best_rating_per_price)
```

```python
price_extremes = reviews.groupby('variety')['price'].agg([min,max])
print(price_extremes)
```

```python
sorted_varieties= price_extremes.sort_values(by=['min','max'], ascending=False)
print(sorted_varieties)
```

```python
reviewer_mean_ratings = reviews.groupby('taster_name')['points'].mean()
print(reviewer_mean_ratings)
```

```python
country_variety_counts = reviews.groupby(['country','variety']).size().sort_values(ascending=False)
print
```
