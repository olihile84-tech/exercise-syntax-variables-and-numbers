```python
import pandas as pd
pd.set_option('display.max_rows', 5)
print("Setup complete")
```

```python
data = {
     'country': ['Italy', 'France', 'Spain','Australia','US'],
    'variety': ['Merlot', 'Chardonnay','Tempranillo', 'Shiraz', 'Pinot Noir'],
    'points': [90,95,88,96,92],
      'price': [25.0, 40.0, 15.0, 50.0, 35.0],
      'province': ['Tuscany', 'Burgundy', 'Rioja', 'Victoria', 'Oregon']

}
reviews = pd.DataFrame(data)
reviews
```

```python
desc = reviews.country
print(desc)
```

```python
first_country = reviews.country[0]
print(first_country)
```

```python
first_three = reviews.iloc[0:3]
print(first_three)
```

```python
df = reviews.loc[0:2,['country','variety']]
print(df)
```

```python
italian_wines = reviews[reviews.country == 'Italy']
print(italian_wines)
```

```python
top_wines = reviews.loc[reviews.points > 90]
print(top_wines)
```

```python
oceania_wines = reviews[reviews.country.isin(['Australia','France'])]
print(oceania_wines)
```
