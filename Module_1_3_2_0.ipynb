```python
import numpy as np

np.random.seed(7)
trade_pnl = np.random.normal(loc=15, scale=80, size=60)

equity_curve = 10000 + np.add.accumulate(trade_pnl)
```

```python
import numpy as np

running_peak = np.maximum.accumulate(equity_curve)
drawdown = equity_curve - running_peak
```

```python
import numpy as np
strategy_pnls = np.random.normal(loc=10, scale=50, size=(60, 3))

equity_table = 10000 + np.add.accumulate(strategy_pnls, axis=0)
```

```python
import numpy as np

returns_pct = np.random.normal(loc=0.002, scale=0.01, size=60)

win_rate = np.sum(returns_pct>0)/ len(returns_pct)* 100
compounded_growth = np.multiply.accumulate(1 + returns_pct)
```
