```python
import numpy as np

A = np.array([[1,1],
              [0, 1]])
B = np.array([[2, 0],
              [3, 4]])

print("Elementwise * :")
print(A * B)

print("\nMatrix product @:")
print(A @ B)

print("\nMatrix product .dot():")
print(A.dot(B))
```

```python
import numpy as np

pnl = np.array([150.0, -50.0, 200.0, -25.0, 300.0, -100.0])

print("ALL PnL:", pnl)
print()
print("Total Profit:", pnl.sum())
print("Worst Trade:", pnl.min())
print("Best Trade:", pnl.max())
print("Average Trade:", pnl.mean())
```

```python
import numpy as np

pnl_table = np.array([
    [150.0, -50.0, 200.0],
     [-25.0, 300.0, -100.0]
])

print("Sum per session (axis=1):")
print(pnl_table.sum(axis=1))


print("\nBest trade per round (axis=0):")
print(pnl_table.sum(axis=0))

print("\nWorst trade per round (axis=0):")
print(pnl_table.min(axis=0))
```

```python
import numpy as np

daily_pnl = np.array([150, -50, 200, -25, 300, -100])
equity_curve = np.cumsum(daily_pnl)

print("Daily PnL:", daily_pnl)
print("Equity curve:", equity_curve)
```
