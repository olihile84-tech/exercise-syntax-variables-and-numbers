```python
import pandas as pd

data1 = {
    'MeetID': [1, 2, 3],
    'MeetName': ['National Championship', 'World Cup', 'Regional Meet'],
    'country': ['Italy', 'France', 'Spain'],
    'variety': ['Merlot', 'Chardonnay', 'Tempranillo'],
    'points': [90, 95, 88],
    'price': [25.0, 40.0, 15.0],
    'region_1': ['Tuscany', 'Burgundy', 'Rioja'],
    'region_2': ['Central Italy', 'Eastern France', None]
}

data2 = {
    'MeetID': [1, 2, 3],
    'AthleteID': [101, 102, 103],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'TotalKg': [300.0, 350.0, 280.0]
}

reviews = pd.DataFrame(data1)
powerlifting_meets = pd.DataFrame(data1)
powerlifting_competitors = pd.DataFrame(data2)
reviews
```

```python
renamed = reviews.rename(columns={'region_1':'region','region_2':'locale'})
print(renamed)
```

```python
reindexed = reviews.rename_axis('wines', axis='rows')
print(reindexed)
```

```python
powerlifting_meets = powerlifting_meets.set_index(['MeetID']).join(powerlifting_competitors.set_index(['MeetID']))
print(powerlifting_meets)
```
