```python
import pandas as pd

data = {
    'country': ['Italy', 'France', 'Spain', 'Australia', 'US'],
        'variety': ['Merlot', 'Chardonnay', 'Tempranillo', 'Shiraz', 'Pinot Noir'],
            'points': [90, 95, 88, 96, 92],
                'price': [25.0, 40.0, 15.0, 50.0, 35.0],
                    'province': ['Tuscany', 'Burgundy', 'Rioja', 'Victoria', 'Oregon']
}
reviews = pd.DataFrame(data)
reviews
```

```python
mean_points = reviews.points.mean()
median_points = reviews.points.median()
print(mean_points, median_points)
```

```python
unique_countries = reviews.country.nunique()
print(unique_countries)
```

```python
country_counts = reviews.country.value_counts()
print(country_counts)
```

```python
mean_price_top = reviews[reviews.points>90]['price'].mean()
print(mean_price_top)
```

```python
expensive_ratio =(len(reviews[reviews.price >30]) /len(reviews)) *100
print(expensive_ratio)
```

```python
noir_wines = reviews[reviews.variety.str.contains('Noir')]
print(noir_wines)
```

```python
special_wines = reviews[
    ((reviews.country=='Italy')&(reviews.points>=89))|
    ((reviews.country=='Australia')&(reviews.points>60))
]
print(special_wines)
```
