```python
import pandas as pd
import numpy as np
firm_data = {
    'Trade_ID': [101, 102, 103, 104, 105, 106, 107, 108],
    'Asset': ['XAUUSD', 'GBPUSD', 'XAUUSD', 'EURUSD', 'XAUUSD', 'GBPUSD', 'EURUSD', 'XAUUSD'],
    'Session': ['London', 'London', 'New_York', 'London', 'New_York', 'Asian', 'New_York', 'London'],
    'Lot_Size':[0.10, 0.05, 0.20, 0.07, 0.15, 0.05, 0.10, 0.12],
    'Pips_Gained': [45, -20, 80, 15, -35, 10, -12, 50],
    'Net_Profit_USD': [450.00, -100.00, 1600.00, 105.00, -525.00, 50.00, -120.00, 500.00]
}

acuity_df = pd.DataFrame(firm_data)
print("---TEST DATA INGESTED SUCCESDDFULLY---")
```

```python

filtered = acuity_df[(acuity_df['Net_Profit_USD']>100)& (acuity_df['Asset']=='XAUUSD')]
print(filtered)
```

```python
session_profit = acuity_df.groupby('Session')['Net_Profit_USD'].sum()
print(session_profit)
```
