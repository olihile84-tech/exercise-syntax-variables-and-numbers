```python
def calculate_position_size(account_balance):
  risk_percentage = 0.25
  stop_loss_pips = 50.0
  pip_value_standard_lot = 10.0
  risk_amount = account_balance *(risk_percentage/100)
  calculated_lot_size = risk_amount/(stop_loss_pips*pip_value_standard_lot)
  if calculated_lot_size < 0.01:
    print("WARNING: Position size too small for standard execution. Adjust Stop Loss parameters")
  else:
    print("---ACUITY EDGE TRADING|SYSTEM REPORT---")
    print("Operating Under:Belial")
    print("Cash at Risk:$"+str(risk_amount))
    print("Recommended Execution Lot Size:"+str(round(calculated_lot_size,2)))
```

```python
calculate_position_size(5000.00)
```
