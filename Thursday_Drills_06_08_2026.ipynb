```python
import numpy as np

a = np.arange (1, 21)

grid = a.reshape(4, 5)

grid[grid % 3 == 0] = -1

print(grid)
```

```python
import numpy as np
import matplotlib.pyplot as plt

current_equity = 90
peak_equity = 120

drawdown_pct = (current_equity - peak_equity)/ peak_equity
print(f"{drawdown_pct:.2%}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(50)
strategy_A = np.random.randn(50).cumsum()
strategy_B = np.random.randn(50).cumsum()

plt.plot(strategy_A, 'r--', label='Strategy A')
plt.plot(strategy_B, 'b-', label='Strategy B')

```
